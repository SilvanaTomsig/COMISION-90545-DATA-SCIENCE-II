# Ejemplo 2 · Resolución (docente)

```python
import pandas as pd
ventas = pd.DataFrame({
    "categoria": ["Electrónica"]*4 + ["Ropa"]*4,
    "producto": ["Celular","Tablet","Auriculares","Cargador","Campera","Jeans","Remera","Medias"],
    "ventas": [900,700,700,200,500,400,400,100]})
ventas
```

## a) Ranking
- `rn`: `cumcount` tras ordenar (desempata por orden de aparición)
- `rnk`: `method='min'` → salta posiciones tras un empate
- `drnk`: `method='dense'` → no salta

```python
g = ventas.groupby('categoria')['ventas']
ventas['rn'] = (ventas.sort_values('ventas', ascending=False, kind='stable')
                .groupby('categoria').cumcount() + 1)
ventas['rnk'] = g.rank(method='min', ascending=False).astype(int)
ventas['drnk'] = g.rank(method='dense', ascending=False).astype(int)
ventas
```

## b) Porcentaje de la categoría

```python
ventas['pct'] = (100 * ventas['ventas'] / g.transform('sum')).round(1)
ventas
```

## c) Top 2 por categoría
Salen 3 filas por categoría porque el empate en el puesto 2 obliga a incluir ambos productos.

```python
top2 = (ventas[ventas['drnk'] <= 2]
        .sort_values(['categoria','drnk'])[['categoria','producto','ventas','drnk']])
top2
```
